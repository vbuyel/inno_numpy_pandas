**2. What is the average age of men (attribute *age*) across the entire dataset?**

```python
avg_age_men = data[data['sex'] == 'Male']['age'].mean()
print(f"Average age of men: {avg_age_men:.2f}")
```

**3. What is the proportion of United States citizens (*native-country* attribute)?**

```python
prop_us_citizens = data[data['native-country'] == 'United-States'].shape[0] / data.shape[0]
print(f"Proportion of United States citizens: {prop_us_citizens:.2%}")
```

**4-5. Calculate the mean and standard deviation of the age of those who earn more than 50K per year (the *salary* attribute) and those who earn less than 50K per year**

```python
# filtered_df = data[data['salary'] == '>50K']
# mean_age_high_earners = filtered_df['age'].mean()
# std_age_high_earners = filtered_df['age'].std()

stats_high_earners = data.groupby('salary')['age'].agg(['mean', 'std'])

print(f"Mean age of high earners: {stats_high_earners.loc['>50K', 'mean']:.2f}")
print(f"Standard deviation of age of high earners: {stats_high_earners.loc['>50K', 'std']:.2f}")

filtered_df = data[data['salary'] == '<=50K']
mean_age_low_earners = filtered_df['age'].mean()
std_age_low_earners = filtered_df['age'].std()

print(f"Mean age of low earners: {mean_age_low_earners:.2f}")
print(f"Standard deviation of age of low earners: {std_age_low_earners:.2f}")
```

**6. Is it true that people who earn more than 50k have at least a higher education? (*education indicator – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate*)**

```python
higher_education = ['Bachelors', 'Prof-school', 'Assoc-acdm', 'Assoc-voc', 'Masters', 'Doctorate']

high_earners_education = data.loc[data['salary'] == '>50K', 'education']
print(high_earners_education.isin(higher_education).all())
```

**7. Find age statistics for each race (*race* attribute) and each gender. Use *groupby* and *describe*. Find the maximum age of males of the *Asian-Pac-Islander* race.**

```python
data_stats = data.groupby(['race', 'sex'])['age'].describe()

max_age = data_stats.loc[('Asian-Pac-Islander', 'Male'), 'max']
print(f"Maximum age of Asian-Pac-Islander males: {int(max_age)}")

data_stats
```

**8. Among whom is the proportion of high-earning individuals (>50K) higher: married or single men (*marital-status* indicator)? We consider those whose *marital-status* begins with *Married* (Married-civ-spouse, Married-spouse-absent, or Married-AF-spouse) to be married; the rest are considered single.**

```python
men = data[data['sex'] == 'Male']
married_men = men[men['marital-status'].str.startswith('Married')]
single_men = men[~men['marital-status'].str.startswith('Married')]

married_proportion = (married_men['salary'] == '>50K').mean()
single_proportion = (single_men['salary'] == '>50K').mean()

print(f"Share of married men earning >50K: {married_proportion:.2%}")
print(f"Share of single men earning >50K: {single_proportion:.2%}")
print("Higher share among:", "married" if married_proportion > single_proportion else "single")
```

**9. What is the maximum number of hours a person works per week (hours-per-week indicator)? How many people work that many hours, and what percentage of them earn a lot?**

```python
max_hours = data['hours-per-week'].max()
print(f"Maximum number of hours a person works per week: {max_hours}")

max_hours_workers = data[data['hours-per-week'] == max_hours]
print(f"Number of people who work {max_hours} hours per week: {max_hours_workers.shape[0]}")

high_earners_proportion = (max_hours_workers['salary'] == '>50K').mean()
print(f"Share of them earning >50K: {high_earners_proportion:.2%}")


data.groupby('salary')['hours-per-week'].describe()
```

**11.Group people into age groups *young*, *adult*, *retiree*, where:**
* *young* corresponds to 16-35 years old
* *adult* - 35-70 years old
* *retiree* - 70-100 years old

**Enter the name of the appropriate age group for each person in the new AgeGroup column.**

```python
data['AgeGroup'] = pd.cut(
    data['age'],
    bins=[16, 35, 70, 100],
    labels=['young', 'adult', 'retiree'],
)
pd.DataFrame(data.groupby('AgeGroup', observed=True)['salary'].value_counts())
```

**12-13. Determine the number of people earning >50K in each age group (AgeGroup column), and also output the name of the age group in which people most often earn more than 50K (>50K)**

```python
salary_counts = data[data['salary'] == '>50K'].groupby('AgeGroup', observed=True)['salary'].count().rename('count')

print(f"Age group with the highest number of people earning >50K: {salary_counts.idxmax()}")

pd.DataFrame(salary_counts)
```

**14. Group people by employment type (the occupation column) and determine the number of people in each group. Then write a filter function, filter_func, that will return only those groups in which the average age (the age column) is 40 or less and in which all employees work more than 5 hours per week (the hours-per-week column).**

```python
grouped_data = data.groupby('occupation').size().rename('count')
pd.DataFrame(grouped_data)
```

```python
def filter_func(group: pd.DataFrame):
    return (group['age'].mean() <= 40) and (group['hours-per-week'] > 5).all()

filtered_data = data.groupby('occupation')[['occupation', 'age', 'hours-per-week']].filter(filter_func)
filtered_data
```
